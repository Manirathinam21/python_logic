```python
from dataclasses import dataclass
```

```python
@dataclass
class volume:
    radius: int
    height: int
    pi:float = 3.14
        
    def cylinder(self):
        out= self.pi*(self.radius**2)*self.height
        return out
    
    def cone(self):
        out= 1/3*self.pi*(self.radius**2)*self.height
        return out
```

```python
vol1=volume(3,6).cylinder()
vol1
```

```python
vol2=volume(3,6).cone()
vol2
```

## Encapsulation

- In encapsulation input arguments of class is private, we can't modify the values of input arguments with help of object.
- Normally in OOPs we can modify the values of input arguments of class through object by calling class name.variable modify value
- In encapsulation even though input arguments are private, we can see the arguments value by 'obj._class__args'

```python
class bank_acct:
    def __init__(self, name, acct_no, balance):
        self.__balance:float = balance
        self.__name:str = name
        self.__acct_no:int = acct_no

    def show_balance(self):
        return (f"your current balance:{self.__balance}")

    def deposit(self, amount):
        self.__balance = self.__balance + amount
        return (f'Amount deposited successfully, your balance: {self.__balance}')

    def withdraw(self, amount):
        if amount <= self.__balance: 
            self.__balance = self.__balance - amount
            return (f'Amount withdrawed, your current balance: {self.__balance}')

    def personal_details(self):
        print (f'Name:{self.__name} \nAcct_No:{self.__acct_no} \nBalance:{self.__balance}')
```

```python
sbi=bank_acct('mani',1913,10000)
```

## Inheritance

```python
class car1:
    def __init__(self, make, manuf, year):
        self.make= make
        self.manuf= manuf
        self.year= year

    def car_details(self):
        return (f"car company is:{self.make}, car name is:{self.manuf}, it's made in {self.year}")


class car2(car1):
    def __init__(self, varient, drive):
        self.varient= varient
        self.drive= drive

    def car_varients(self):
        return (f"car is a {self.varient}, {self.drive} model")

class car3(car2):
    def review(self):
        return ('it a very good car with highest safety rating of NCAP:5')
```

```python
tata_car1= car1('tata','safari','2023')
tata_car2= car2('petrol', '4w')
```

## Decorator


```python
import time

def decorator(func):
    def decorator_wrapper(num):
        start= time.time()
        print(f"Checking if {num} is a prime number...")
        result=func(num)
        print(result)
        print(f'Given function stopped')
        end= time.time()
        print(f'Time taken:{start-end}')
    return decorator_wrapper
```

```python
@decorator
def check_prime_num(num):
    if num > 1:
        for i in range(2, num):
            if num % i == 0:
                return f'Given {num} is not a prime number'
        else:
            return f'Given {num} is a prime number'
    else:
        return f'Given {num} is not a prime number'

check_prime_num(22)
```

### Class- Static Method

```python
class maths:
    def __init__(self):
        self.pi= 3.14

    def vol_cylinder(self, radius, height):
        return (self.pi*radius**2*height)

    @staticmethod
    def vol_cone(radius, height):
        return (1/3*3.14*radius**2*height)
```

```python
cylind= maths()
```

```python
cylind.__dict__
```

```python
cylind.vol_cylinder(3,4)
```

```python
'''if we use staticmethod decorator for properties or methods in class, 
we can directly access the method and pass values into it. instead of assigning it to object and then call it use the methods
we don't need to mention self in function arguments, just pass arguments when using static decorator. it reduces the memory consumption 
'''

maths().vol_cone(2,3)
```

## Interview qstn Input='My Name is Lisa' expected Output= 'as iLsi em aNyM'

```python
def reverse_string_keep_spaces(s):
    space_index= [i for i, ch in enumerate(s) if ch==' ']
    
    str= [ch for ch in s if ch!=' ']
    rev_str= str[::-1]
    
    result=[]
    
    index=0
    for idx in range(len(s)):
        if idx in space_index:
            result.append(' ')
        else:
            result.append(rev_str[index])
            index+=1
    
    return ''.join(result)

s= 'My Name is Lisa'
reverse_string_keep_spaces(s)
```

## Interview QSTN

Input: nums= [10,9, 2,5,3,7,101,18]

write a python function that returns the length of the longest strictly increasing sequence of nums answers is 4

Output: 4 (2,3,7,18)

```python
from bisect import bisect_left

def length_of_lis(nums):
    sub=[]
    for num in nums:
        i=bisect_left(sub, num)
        if i == len(sub):
            sub.append(num)
        else:
            sub[i]=num
    return len(sub), sub
```

```python
nums= [10,9, 2,5,3,7,101,18]
length_of_lis(nums)
```

## Interview question- Vowels count in a sentence

```python
def vowel_count(text):
    vowel=['a','i','o','e','u']
    count = {}.fromkeys(vowel, 0)
    
    for char in text:
        if char in count:
            count[char]+=1
    return count

if __name__=="__main__":
    text=input("Enter text : ")
    print(vowel_count(text))
```

### Fibonacci function

```python
def fibonacci_func(n):
    a,b=0,1
    for x in range(n):
        yield a
        a,b = b, a+b        
```

```python
# one approach to see output

for x in fibonacci_func(12):
    print(x)
```

```python
# second approach to see output
out= list(fibonacci_func(10))
out
```
