## DateTime Format

```python
from datetime import datetime

cur_time=datetime.now()
cur_time
```

```python
print(cur_time)
```

```python
type(cur_time)
```

### Strptime() --> convert string to  datetime object

```python
time='2023-07-25 10:12:56'
strptime_out=datetime.strptime(time, "%Y-%m-%d %H:%M:%S")
strptime_out
```

```python
type(strptime_out)
```

### strftime() --> convert datetime object to string 

```python
strftime_out=datetime.strftime(strptime_out, "%Y-%m-%d %H:%M:%S")
strftime_out
```

```python
type(strftime_out)
```

**Interchange the date and month position in the output**

```python
datetime.strftime(strptime_out, "%Y-%d-%m %H:%M:%S")
```

**Get only last two digits of year**

```python
datetime.strftime(strptime_out, "%y-%m-%d %H:%M:%S")
```

**Get month name in date instead of month number in date**

```python
datetime.strftime(strptime_out, "%d-%h-%Y %H:%M:%S")
```

**To get Only date part**

```python
datetime.strftime(strptime_out, "%D")
```

**To get only Year**

```python
datetime.strftime(strptime_out, "%Y")
```

**To get only Month number**

```python
datetime.strftime(strptime_out, "%m")
```

**To get only Month Name**

```python
datetime.strftime(strptime_out, "%h")
```

**To get only full Month Name**

```python
datetime.strftime(strptime_out, "%B")
```

**To get the day name**

```python
datetime.strftime(strptime_out, "%A")
```

**To get the day name in short**

```python
datetime.strftime(strptime_out, "%a")
```

**To get only date**

```python
datetime.strftime(strptime_out, "%d")
```

**To get date in YYYY-DD--MM format**

```python
time_now=datetime.now()
now_out=datetime.strftime(time_now, "%F")
now_out
```

**To get Time in HH:MM:SS format**

```python
datetime.strftime(strptime_out, "%T")
```

**To get Time in HH:MM format**

```python
datetime.strftime(strptime_out, "%R")
```

**To get Time of day (AM or PM)**

```python
datetime.strftime(strptime_out, "%p")
```

**To get complete Time of day (HH:MM:SS AM or PM)**

```python
datetime.strftime(strptime_out, "%r")
```

### Add & subtract date time in python timedelta

```python
from datetime import timedelta
timedelta(days=0, seconds=0, microseconds=0)
```

```python
time_output=datetime(2023, 7, 25, 20, 30, 8)
time_output
```

**Add 2 days  and 1 hour**

```python
time_output+timedelta(days=2, hours=1)
```

**subtract 1 day and 30 mins**

```python
time_output-timedelta(days=1, minutes=30)
```
